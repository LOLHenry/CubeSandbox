# CubeSandbox 本机演示（鲲鹏）

Notebook 与 CubeAPI 都在同一台鲲鹏上运行。

使用前：`cp .env.example .env`，填写 `CUBE_TEMPLATE_ID`；确认 `curl http://127.0.0.1:3000/health` 正常。

## Cell 1 — 加载配置

```python
import os
from pathlib import Path
from dotenv import load_dotenv
from e2b_code_interpreter import Sandbox

load_dotenv(Path(".env"), override=True)

for k in ["E2B_API_URL", "E2B_API_KEY", "CUBE_TEMPLATE_ID", "SSL_CERT_FILE"]:
    print(f"{k} = {os.environ.get(k)}")
```

## Cell 2 — 创建沙箱（本进程内只 create 一次）

```python
sbx = Sandbox.create(template=os.environ["CUBE_TEMPLATE_ID"], timeout=600)
print("sandbox_id =", sbx.sandbox_id)
```

## Cell 3 — 架构 / Python 版本

```python
r = sbx.run_code("import platform, sys\nprint(platform.machine())\nprint(sys.version)")
print(r)
```

## Cell 4 — 探测数据科学包

```python
r = sbx.run_code(
    "import importlib.util as u\n"
    "for m in ['numpy','pandas','matplotlib','sklearn']:\n"
    "    print(m, bool(u.find_spec(m)))\n"
)
print(r)
```

## Cell 5 — 简单计算

```python
r = sbx.run_code("print(sum(range(1, 101)))")
print(r)
```

## Cell 6 — Shell（本机 DNS 正常时可用）

```python
print(sbx.commands.run("uname -a").stdout)
print(sbx.commands.run("df -h / | tail -1").stdout)
```

## Cell 7 — 销毁沙箱

```python
sbx.kill()
print("sandbox destroyed")
```
